### 4. 裂变碎片属性分布
<a id='ffHistograms'></a>

使用 Matplotlib 绘制裂变碎片各项属性的分布。

````python
# 绘制裂变碎片属性分布
A = hist.getA()  # 获取所有碎片的质量数 A
AL = hist.getALF()  # 获取轻碎片的质量数 A
AH = hist.getAHF()  # 获取重碎片的质量数 A
fig = plt.figure(figsize=(8,6))
bins = np.arange(min(A),max(A))
h,b = np.histogram(A,bins=bins,density=True)
plt.plot(b[:-1],h,'-o')
plt.xlabel('质量数 (A)')
plt.ylabel('概率密度')
plt.show()

Z = hist.getZ()
ZL = hist.getZLF()
ZH = hist.getZHF()
fig = plt.figure(figsize=(8,6))
bins = np.arange(min(Z),max(Z))
h,b = np.histogram(Z,bins=bins,density=True)
plt.plot(b[:-1],h,'-o')
plt.xlabel('电荷数 (Z)')
plt.ylabel('概率密度')
plt.show()

fig = plt.figure(figsize=(8,6))
TKEpre = hist.getTKEpre()  # 中子发射前的总动能
TKEpost = hist.getTKEpost()  # 中子发射后的总动能
bins = np.arange(min(TKEpre),max(TKEpre))
h,b = np.histogram(TKEpre,bins=bins,density=True)
plt.plot(0.5*(b[:-1]+b[1:]),h,'-o',label='中子发射前')
bins = np.arange(min(TKEpost),max(TKEpost))
h,b = np.histogram(TKEpost,bins=bins,density=True)
plt.plot(0.5*(b[:-1]+b[1:]),h,'-o',label='中子发射后')
plt.legend()
plt.xlabel('总动能 (MeV)')
plt.ylabel('概率密度')
plt.show()
````

二维直方图可以展示 CGMF 计算得到的不同物理量之间的相关性。

````python
TKEpre = hist.getTKEpre()
TXE = hist.getTXE()
bx = np.arange(min(TKEpre),max(TKEpre))
by = np.arange(min(TXE),max(TXE))
fig = plt.figure(figsize=(8,6))
plt.hist2d(TKEpre,TXE,bins=(bx,by),density=True)
plt.xlabel('总动能 (MeV)')
plt.ylabel('总激发能 (MeV)')
plt.colorbar()
plt.show()
````

### 5. 相关观测量
<a id='correlations'></a>

裂变过程中许多观测量彼此相关。有时适合用二维直方图展示（例如上方的 TKE-TXE 图），有时则适合将观测量表示为质量数的函数。CGMFtk 提供了相应的计算函数，下面给出示例：

````python
# 平均中子数及总动能随碎片质量数的变化
# 平均伽马数、激发能、中子发射前的动能和自旋也可表示为质量数的函数
nubarA = hist.nubarA()
TKEA = hist.TKEA()
fig = plt.figure(figsize=(16,6))
plt.subplot(121)
plt.plot(nubarA[0],nubarA[1],'ko')
plt.xlabel('质量数 (u)')
plt.ylabel(r'平均中子数 $\overline{\nu}$')
plt.subplot(122)
plt.plot(TKEA[0],TKEA[1],'ko')
plt.xlabel('质量数 (u)')
plt.ylabel('总动能 (MeV)')
plt.tight_layout()
plt.show()
````

### 7. 伽马属性
<a id='gammas'></a>

````python
# 计算并绘制伽马射线多重数分布
nug,pnug = hist.Pnug()
# 设置 100 keV（0.1 MeV）的能量阈值
nugth,pnugth = hist.Pnug(Eth=0.1)
fig = plt.figure(figsize=(8,6))
plt.plot(nug,pnug,'k*--',markersize=10,label='能量阈值=0 MeV')
plt.plot(nugth,pnugth,'ro:',markersize=10,label='能量阈值=100 keV')
plt.xlabel(r'伽马数 $N_\gamma$')
plt.ylabel(r'概率 $P(N_\gamma)$')
plt.legend()
plt.show()
````

````python
# 计算并绘制瞬发裂变伽马射线谱
fig = plt.figure(figsize=(16,6))
plt.subplot(121)
ebins,pfgs = hist.pfgs()
plt.step(ebins,pfgs,where='mid')
plt.xlim(0,5)
plt.xlabel(r'出射伽马射线能量 (MeV)')
plt.ylabel('瞬发裂变伽马射线谱 (PFGS)')
plt.subplot(122)
plt.step(ebins,pfgs,where='mid')
plt.xlim(0.1,5)
plt.xscale('log')
plt.yscale('log')
plt.xlabel(r'出射伽马射线能量 (MeV)')
plt.ylabel('瞬发裂变伽马射线谱 (PFGS)')
plt.ylim(1e-2,30)
plt.tight_layout()
plt.show()
````

````python
# 瞬发伽马射线的平均数
print('平均伽马射线数（每次裂变）=', hist.nubargtot())
print('每个裂变碎片的平均伽马射线数=', hist.nubarg())
# 设置能量阈值 Eth（单位：MeV）后，计算瞬发伽马射线的平均数
Enth = 0.1  # 100 keV（0.1 MeV）
print('平均伽马射线数（每次裂变）=', hist.nubargtot(Eth=Enth))
print('每个裂变碎片的平均伽马射线数=', hist.nubarg(Eth=Enth))
````

````python
# 进行伽马射线能谱分析
gE = 0.2125  # 能量为 212.5 keV 的伽马射线
dE = 0.01  # 能量分辨率为 1%
gspec1 = hist.gammaSpec(gE,dE*gE,post=True)

# 计算各质量数 A 和电荷数 Z 对应的事件比例
As1 = np.unique(gspec1[:,1])
totEvents = len(gspec1)
fracA1 = []
for A in As1:
    mask = gspec1[:,1]==A
    fracA1.append(len(gspec1[mask])/totEvents)
Zs1 = np.unique(gspec1[:,0])
fracZ1 = []
for Z in Zs1:
    mask = gspec1[:,0]==Z
    fracZ1.append(len(gspec1[mask])/totEvents)

fig = plt.figure(figsize=(8,6))
plt.plot(As1,fracA1,'--')
plt.xlabel('裂变碎片质量数 (A)')
plt.ylabel('事件比例')
plt.text(135,0.150,r'伽马能量 $E_\gamma$=212.5 keV',fontsize=18)
plt.show()

fig = plt.figure(figsize=(8,6))
plt.plot(Zs1,fracZ1,'--')
plt.xlabel('裂变碎片电荷数 (Z)')
plt.ylabel('事件比例')
plt.show()
````

````python
# 伽马射线的平均能量
print('实验室系伽马射线能量：')
print('所有伽马射线的平均能量=', hist.meanGammaElab())
print('来自轻碎片的伽马射线平均能量=', hist.meanGammaElabLF())
print('来自重碎片的伽马射线平均能量=', hist.meanGammaElabHF())
````

注意：当前版本的 CGMF 在历史文件中只记录经过多普勒频移的实验室系伽马能量。

### 8. 伽马射线时间信息
<a id='timing'></a>

还可以分析较晚发射的瞬发伽马射线随时间的变化。在运行 CGMF 时加入 `-t -1` 选项，伽马射线发射时间便会写入 CGMF 历史文件。`Histories` 类可根据历史文件头读取这些时间信息。

````python
histTime = fh.Histories(workdir + timeFile)
````

````python
# 获取伽马射线的发射时间
gammaAges = histTime.getGammaAges()
````

`nubarg()` 或 `nubargtot()` 可以计算平均伽马射线多重数随时间的变化。设置 `timeWindow=True` 时使用函数内置的默认时间点；也可以向 `timeWindow` 传入 NumPy 数组或列表来指定时间点。还可以通过 `Eth` 设置伽马射线能量下限。

````python
times,nubargTime = histTime.nubarg(timeWindow=True)  # timeWindow 可设为布尔值或以秒为单位的时间列表
fig = plt.figure(figsize=(8,6))
plt.plot(times,nubargTime,'o',label='能量阈值=0 MeV')
times,nubargTime = histTime.nubarg(timeWindow=True,Eth=0.1)
plt.plot(times,nubargTime,'o',label='能量阈值=0.1 MeV')
plt.xlabel('裂变后的时间 (s)')
plt.ylabel('平均伽马射线多重数')
plt.xscale('log')
plt.legend()
plt.show()
````

瞬发裂变伽马射线谱函数 `pfgs()` 也可以计算指定时间窗口内的能谱。使用 `minTime` 和 `maxTime` 分别设置时间窗口的起始和结束时间。

````python
fig = plt.figure(figsize=(8,6))
bE,pfgsTest = histTime.pfgs(minTime=5e-8,maxTime=500e-8)
plt.step(bE,pfgsTest,label='指定时间窗口')
bE,pfgsTest = histTime.pfgs()
plt.step(bE,pfgsTest,label='所有事件')
plt.yscale('log')
plt.xlim(0,2)
plt.ylim(0.1,100)
plt.xlabel('伽马射线能量 (MeV)')
plt.ylabel('瞬发裂变伽马射线谱')
plt.legend()
plt.show()
````

````python
# 计算特定裂变碎片的伽马射线多重数随裂变后时间的变化
times,gMultiplicity = histTime.gammaMultiplicity(minTime=1e-8,maxTime=1e-6,Afragment=134,Zfragment=52)

# 与该能级半衰期对应的指数衰减曲线进行比较
f = np.exp(-times*np.log(2)/1.641e-7)  # 134Te 的半衰期为 164.1 ns
norm = gMultiplicity[0]/f[0]

fig = plt.figure(figsize=(8,6))
plt.plot(times*1e9,gMultiplicity/norm,'k-',label='CGMF 计算')
plt.plot(times*1e9,f,'r--',label=r'指数衰减（半衰期 $\tau_{1/2}$）')
plt.legend()
plt.yscale('log')
plt.xlabel('裂变后的时间 (ns)')
plt.ylabel('伽马射线多重数（任意单位）')
plt.show()
````

````python
# 计算特定原子核能级的同质异能素比
# 例如：99Nb 中 1/2- 能级；基态为 9/2+，寿命为 150 s
r = histTime.isomericRatio(thresholdTime=1,A=99,Z=41,Jm=0.5,Jgs=4.5)
print('99Nb：', round(r,2))
# 例如：133Te 中 11/2- 能级；基态为 3/2+，寿命为 917.4 s
r = histTime.isomericRatio(thresholdTime=1,A=133,Z=52,Jm=5.5,Jgs=1.5)
print('133Te：', round(r,2))
````

### 9. 角关联
<a id='angles'></a>

计算裂变碎片相对于束流轴（z 轴）的角分布时，可以通过 `afterEmission=True/False` 选择使用中子发射后的角度（`True`）或发射前的角度（`False`）。默认值为 `True`。

````python
# 计算裂变碎片与 z 轴（束流轴）夹角的余弦
FFangles = hist.FFangles()
bins = np.linspace(-1,1,30)
h,b = np.histogram(FFangles,bins=bins,density=True)
# 仅统计轻碎片
hLight,b = np.histogram(FFangles[::2],bins=bins,density=True)
# 仅统计重碎片
hHeavy,b = np.histogram(FFangles[1::2],bins=bins,density=True)
````

````python
x = 0.5*(b[:-1]+b[1:])
fig = plt.figure(figsize=(8,6))
plt.plot(x,h,'k*',label='所有碎片')
plt.plot(x,hLight,'ro',label='轻碎片')
plt.plot(x,hHeavy,'b^',label='重碎片')
plt.xlabel(r'$\cos(\theta)$')
plt.ylabel('概率密度')
plt.ylim(0.45,0.55)
plt.title('裂变碎片相对于束流轴的角分布')
plt.legend()
plt.show()
````

计算中子相对于束流轴（z 轴）的角度时，可以用 `Eth` 设置中子能量阈值（单位：MeV）；`lab=True`（默认值）表示实验室系，`lab=False` 表示复合系统质心系。`includePrefission=True`（默认值）表示包含裂变前中子，设为 `False` 则不包含。裂变前中子只支持在实验室系中计算。

````python
# 计算中子与 z 轴（束流轴）之间夹角的余弦
nAllLab,nLLab,nHLab = hist.nangles(lab=True)  # 所有中子、来自轻碎片的中子、来自重碎片的中子
nAllCM,nLCM,nHCM = hist.nangles(lab=False)  # 复合系统质心系
````

````python
bins = np.linspace(-1,1,30)
hAllLab,b = np.histogram(nAllLab,bins=bins,density=True)
hLightLab,b = np.histogram(nLLab,bins=bins,density=True)
hHeavyLab,b = np.histogram(nHLab,bins=bins,density=True)
hAllcm,b = np.histogram(nAllCM,bins=bins,density=True)
hLightcm,b = np.histogram(nLCM,bins=bins,density=True)
hHeavycm,b = np.histogram(nHCM,bins=bins,density=True)
````

````python
x = 0.5*(b[:-1]+b[1:])
fig = plt.figure(figsize=(8,6))
plt.plot(x,hAllLab,'k*',label='所有中子')
plt.plot(x,hLightLab,'ro',label='来自轻碎片')
plt.plot(x,hHeavyLab,'b^',label='来自重碎片')
plt.xlabel(r'$\cos(\theta)$')
plt.ylabel('概率密度')
plt.ylim(0.45,0.55)
plt.title('实验室系中子相对于束流轴的角分布')
plt.legend()
plt.show()
````

````python
fig = plt.figure(figsize=(8,6))
plt.plot(x,hAllLab,'k*',label='实验室系')
plt.plot(x,hAllcm,'ro',label='复合系统质心系')
plt.xlabel(r'$\cos(\theta)$')
plt.ylabel('概率密度')
plt.ylim(0.45,0.55)
plt.legend()
plt.show()
````

计算所有中子与轻、重裂变碎片之间的夹角时，也可以使用前述选项：`Eth`（中子能量阈值）、`afterEmission`（使用中子发射前或发射后的碎片角度）及 `includePrefission`（是否包含裂变前中子）。

````python
# 计算中子与裂变碎片之间夹角的余弦
nFall,nFLight,nFHeavy = hist.nFangles()
bins = np.linspace(-1,1,30)
hall,b = np.histogram(nFall,bins=bins,density=True)
hlight,b = np.histogram(nFLight,bins=bins,density=True)
hheavy,b = np.histogram(nFHeavy,bins=bins,density=True)
````

````python
x = 0.5*(b[:-1]+b[1:])
fig = plt.figure(figsize=(8,6))
plt.plot(x,hall,'k*',label='所有碎片')
plt.plot(x,hlight,'ro',label='轻碎片')
plt.plot(x,hheavy,'b^',label='重碎片')
plt.xlabel(r'$\cos(\theta)$')
plt.ylabel('概率密度')
plt.legend()
plt.show()
````
